Se requiere calcular el número total de nuevas plantas producidas en un vivero en un mes n siguiendo un modelo de crecimiento basado en la serie de Fibonacci:
Casos base: T(0) = 0 y T(1) = 1.
Relación de recurrencia: T(n) = T(n-1) + T(n-2) para n >= 2.
El cálculo debe realizarse obligatoriamente utilizando programación dinámica con el enfoque de memoización (top-down) para optimizar el rendimiento y evitar recálculos innecesarios.

### 5) Análisis de complejidad **[15 pts]**
Incluye:
- Justificación

Sin memoización: Calcula lo mismo varias veces y se vuelve exponencial.

Con memoización: Guarda y reutiliza resultados en O(1), bajando a lineal O(n).

- Función T[n]

T(n) = c1 si n <= 1 o memo[n] != -1; en otro caso T(n) = T(n-1) + T(n-2) + c2. Como los subproblemas se guardan en memo, la recurrencia se reduce a T(n) = O(n).

## 6) Código funcional **[15 pts]**
Incluye tu implementación completa.

No pegues código incompleto; debe ser ejecutable.


```python
def T(n, memo=None):
    if memo is None:
        memo = [-1] * (n + 1)

    if memo[n] != -1:
        return memo[n]

    if n <= 1:
        return n

    memo[n] = T(n - 1, memo) + T(n - 2, memo)
    return memo[n]


def main():
    n = -1

    while n < 0:
        n = int(input("Digite el mes que desea calcular: "))

    memo = (n + 1) * [-1]
    r = T(n, memo)

    print("En el mes", n, "tendrá", r, "plantas")


# Pruebas
assert T(0) == 0
assert T(1) == 1
assert T(6) == 8

main()

```
