```python
monedas = [1000, 500, 200, 100, 50]


def cambio_minimo(cambio):

    # dp[i] = cantidad mínima de monedas para formar i

    dp = [float('inf')] * (cambio + 1)

    # Guarda qué moneda se utilizó para llegar a cada valor

    usada = [0] * (cambio + 1)

    # Caso base

    dp[0] = 0

    # Programación dinámica

    for valor in range(1, cambio + 1):

        for moneda in monedas:

            if moneda <= valor:

                if dp[valor - moneda] + 1 < dp[valor]:

                    dp[valor] = dp[valor - moneda] + 1

                    usada[valor] = moneda

    # Reconstruir las monedas utilizadas

    cantidad_monedas = {}

    valor = cambio

    while valor > 0:

        moneda = usada[valor]

        if moneda not in cantidad_monedas:

            cantidad_monedas[moneda] = 0

        cantidad_monedas[moneda] += 1

        valor -= moneda

    return dp[cambio], cantidad_monedas


while True:

    entrada = input(

        "Ingrese el valor del cambio "

        "(o 'salir' para terminar): "

    )

    if entrada.lower() == "salir":

        print("Programa finalizado.")

        break

    try:

        cambio = int(entrada)

    except ValueError:

        print("Error: debe ingresar un número entero.\n")

        continue

    if cambio == 0:

        print("Programa finalizado.")

        break

    if cambio < 0:

        print("Error: no se permiten números negativos.\n")

        continue

    cantidad_total, cantidades = cambio_minimo(cambio)

    print("\nDesglose:")

    for moneda in monedas:

        cantidad = cantidades.get(moneda, 0)

        print(f"Monedas de ${moneda}: {cantidad}")

    print("\nCantidad mínima de monedas:", cantidad_total)

    print("-" * 30)
 
```

#### Código base para benchmarking

```python
monedas = [1000, 500, 200, 100, 50]


def cambio_minimo(cambio):

    # dp[i] = cantidad mínima de monedas para formar i
    dp = [float('inf')] * (cambio + 1)

    # Guarda qué moneda se utilizó
    usada = [0] * (cambio + 1)

    # Caso base
    dp[0] = 0

    # Programación dinámica
    for valor in range(1, cambio + 1):

        for moneda in monedas:

            if moneda <= valor:

                if dp[valor - moneda] + 1 < dp[valor]:

                    dp[valor] = dp[valor - moneda] + 1
                    usada[valor] = moneda

    # Reconstruir las monedas utilizadas
    cantidad_monedas = {}

    valor = cambio

    while valor > 0:

        moneda = usada[valor]

        if moneda == 0:
            return None, {}

        if moneda not in cantidad_monedas:
            cantidad_monedas[moneda] = 0

        cantidad_monedas[moneda] += 1

        valor -= moneda

    return dp[cambio], cantidad_monedas


def bench(n: int, step: int = 1) -> list[tuple[int, float]]:

    print("Generando datos ...", end='')

    from random import randint
    import time

    results = []

    for size in range(1, n, step):

        # El cambio siempre es múltiplo de 50
        cambio = randint(1, size) * 50

        inicio = time.perf_counter()

        cambio_minimo(cambio)

        fin = time.perf_counter()

        tiempo = fin - inicio

        results.append((size, tiempo))

    print(" listo.")

    return results


def dibujar(x, y, dimensiones=(7, 5)):

    assert len(x) == len(y)

    import matplotlib.pyplot as plt

    plt.figure(figsize=dimensiones)

    plt.plot(x, y)

    plt.show()


# === LLAMADO ===

tiempos = bench(10, step=1)

x = [size for size, timing in tiempos]

y = [timing for size, timing in tiempos]

dibujar(x, y)
```
