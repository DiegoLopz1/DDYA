MERGE SORT
El Merge Sort o también llamado ordenamiento por mezcla es un algoritmo  de ordenamiento eficiente que utiliza la técnica "divide y vencerás". Divide una lista en parte partes mas pequeñas o también llamados subarreglos para así tener elementos individuales y formar una nueva lista ordenada o encontrar un elemento especifico. La complejidad del Merge Sort es O(n log n)
 
BUSQUEDA BINARIA
La búsqueda binaria es un algoritmo eficiente para encontrar un elemento en una lista ordenada. Funciona comparando el valor buscado con el elemento del medio. Si no coinciden, se descarta la mitad de la lista donde no puede estar el objetivo y se repite el proceso en la otra mitad. La complejidad de la búsqueda binaria es O(log n)
 
FUNCION T(n)
 
T(n) = T(n/2) + O(1)
 
MEJOR Y PEOR CASO 
-MERGE SORT
	Mejor caso: O(n log n)
	Peor caso: O(n log n)
En este caso el mejor y peor caso son iguales porque el Merge Sort realiza el proceso de dividir y combinar aunque la lista ya esté ordenada.
-BUSQUEDA BINARIA
	Mejor caso: O(1)
	Peor caso: O(log n)
El mejor caso ocurre cuando el código que buscamos está justo en la mitad de la lista y se encuentra en la primera comparación. 
El peor caso ocurre cuando se necesitan varias comparaciones para encontrar el código o determinar que no existe.

```python
# Escribe tu código aquí.

#Codigo Merge Sort
def merge_sort(lista):
    if len(lista) <= 1:
        return lista
 
    mitad = len(lista) // 2
 
    izquierda = merge_sort(lista[:mitad])
    derecha = merge_sort(lista[mitad:])
 
    resultado = []
    i = 0
    j = 0
 
    while i < len(izquierda) and j < len(derecha):
        if izquierda[i] < derecha[j]:
            resultado.append(izquierda[i])
            i += 1
        else:
            resultado.append(derecha[j])
            j += 1
 
    while i < len(izquierda):
        resultado.append(izquierda[i])
        i += 1
 
    while j < len(derecha):
        resultado.append(derecha[j])
        j += 1
 
    return resultado
 
 
def main():
    productos = []
 
    n = int(input("Ingrese la cantidad de productos: "))
 
    for i in range(n):
        codigo = int(input("Ingrese el codigo del producto: "))
        productos.append(codigo)
 
    print("Lista original:", productos)
 
    productos = merge_sort(productos)
 
    print("Lista ordenada:", productos)
 
 
main()
```

```python

#Codigo Busqueda Binaria


def merge_sort(lista):
    if len(lista) <= 1:
        return lista
 
    mitad = len(lista) // 2
 
    izquierda = merge_sort(lista[:mitad])
    derecha = merge_sort(lista[mitad:])
 
    resultado = []
    i = 0
    j = 0
 
    while i < len(izquierda) and j < len(derecha):
        if izquierda[i] < derecha[j]:
            resultado.append(izquierda[i])
            i += 1
        else:
            resultado.append(derecha[j])
            j += 1
 
    while i < len(izquierda):
        resultado.append(izquierda[i])
        i += 1
 
    while j < len(derecha):
        resultado.append(derecha[j])
        j += 1
 
    return resultado
 
 
def busqueda_binaria(lista, numero):
    inicio = 0
    final = len(lista) - 1
 
    while inicio <= final:
        medio = (inicio + final) // 2
 
        if lista[medio] == numero:
            return True
 
        if numero < lista[medio]:
            final = medio - 1
        else:
            inicio = medio + 1
 
    return False
 
 
def main():
    productos = []
 
    n = int(input("Ingrese la cantidad de productos: "))
 
    for i in range(n):
        codigo = int(input("Ingrese el codigo del producto: "))
        productos.append(codigo)
 
    print("Lista original:", productos)
 
    productos = merge_sort(productos)
 
    print("Lista ordenada:", productos)
 
    numero = int(input("Ingrese el codigo que desea buscar: "))
 
    if busqueda_binaria(productos, numero):
        print("El producto con codigo", numero, "fue encontrado.")
    else:
        print("El producto con codigo", numero, "no se encuentra registrado.")
 
 
main()
```
